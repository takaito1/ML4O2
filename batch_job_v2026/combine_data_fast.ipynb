```python
import numpy as np
import xarray as xr
from multiprocessing import Pool
```

```python
veralg = '1'
```

```python
def assemble_data(n):
    ds1=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.1.4.6.4.nc')
    ds2=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.2.4.6.4.nc')
    ds3=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.3.4.6.4.nc')
    ds4=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.4.4.6.4.nc')
    ds5=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.5.4.6.4.nc')
    ds6=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.6.4.6.4.nc')
    ds7=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.7.4.6.4.nc')
    ds8=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.8.4.6.4.nc')
    ds9=xr.open_dataset(f'/glade/derecho/scratch/ito/ML4O2_results/O2map_v{veralg}.3.12.4.6.4.nc')
    #
    if n<9:
        n0=n*10
        n1=(n+1)*10
        tmp=np.zeros((9,10,180,360))
        o2 = np.zeros((732,10,180,360))*np.nan
    else:
        n0=n*10
        n1=102
        tmp=np.zeros((9,12,180,360))
        o2 = np.zeros((732,12,180,360))*np.nan
    #
    for t in range(732):
        if t%120 == 0: print(t//12+1965)
        tmp[0,:,:,:]=ds1.o2est[t,n0:n1,:,:].to_numpy()
        tmp[1,:,:,:]=ds2.o2est[t,n0:n1,:,:].to_numpy()
        tmp[2,:,:,:]=ds3.o2est[t,n0:n1,:,:].to_numpy()
        tmp[3,:,:,:]=ds4.o2est[t,n0:n1,:,:].to_numpy()
        tmp[4,:,:,:]=ds5.o2est[t,n0:n1,:,:].to_numpy()
        tmp[5,:,:,:]=ds6.o2est[t,n0:n1,:,:].to_numpy()
        tmp[6,:,:,:]=ds7.o2est[t,n0:n1,:,:].to_numpy()
        tmp[7,:,:,:]=ds8.o2est[t,n0:n1,:,:].to_numpy()
        tmp[8,:,:,:]=ds9.o2est[t,n0:n1,:,:].to_numpy()
        tmp1=np.nanmean(tmp,axis=0)
        tmp2=np.where(tmp1==0,np.nan,tmp1)
        o2[t,:,:,:]=tmp2
    #
    z = ds1.depth[n0:n1]
    x = ds1.lon
    y = ds1.lat
    time = ds1.time
    da=xr.DataArray(data=o2,name='o2',dims=['time','depth','lat','lon'],
                   coords={'time':time,'depth':z,'lat':y,'lon':x})
    ds=da.to_dataset()
    ds.to_netcdf(f'/glade/derecho/scratch/ito/ML4O2_results/o2_temp_{n}.nc')
```

```python
if __name__ == '__main__':
    with Pool(10) as p:
        print(p.map(assemble_data, np.arange(0,10,1)))
#for n in range(10):
#    assemble_data(n)
```
